```python
# recursive problem 

# 5!, 5 * 4 * 3 * 2 * 1
# 6!, 6 * (5 * 4 * 3 * 2 *1) - 6! == 6 * 5! 

def factorial(num: int) -> int:
    '''
    Examples:
        >>> factorial(5!) 
        120
        >>> factorial(4!)
        24
        >>> factorial(1!)
        1
        >>> factorial(0)  # defined by math to be 1
        1
        >>> factorial(-1) # invalid ignore for now
        None
    ''' 
    if num == 1 or num == 0:  # base case
        return 1   
    return num * factorial(num - 1) # recursive element/case
```

```python
print(factorial(5))
```

```python
# recursive problem 

# 5!, 5 * 4 * 3 * 2 * 1
# 6!, 6 * (5 * 4 * 3 * 2 *1) - 6! == 6 * 5! 

def factorial_with_memory(num: int, memory: list = []) -> int:
    '''
    Examples:
        >>> factorial(5!) 
        120
        >>> factorial(4!)
        24
        >>> factorial(1!)
        1
        >>> factorial(0)  # defined by math to be 1
        1
        >>> factorial(-1) # invalid ignore for now
        None
    ''' 
    if num == 1 or num == 0:  # base case
        memory.append(1)
        return 1
       
    answer = num * factorial_with_memory(num - 1, memory) # recursive element/case
    memory.append(answer)
    return answer
```

```python
mem = []
answer = factorial_with_memory(5, mem)
print(f"answer: {answer} with a list of {mem}")
```

```python
# Reverse a string using recursion - keep it simple / don't use built ins 


def reverse(word: str) -> str:
    ''' 
    Examples:
      >>> reverse('Hello')
      'olleH'
      >>> reverse('ada')
      'ada'
      >>> reverse('a')
      'a' 
      >>> reverse('')
      ''
    '''
    if len(word) < 2:
        return word
    
    first_char = word[0]
    rest = word[1:]

    new_word = reverse(rest) + first_char
    return new_word
```

```python
reverse('Hello')
```

```python
# fibonacci  1 1 2 3 5 8  
# (n - 1) + ( n-2 )
## BAD RECURSION - never double or more your function calls (unless that is normal)

def fib(n: int) -> int:
    if n <= 1:
        return 1 
    return fib(n-2) + fib(n-1)
```

```python
def fib_loop(n: int) -> int: 
    count = 1 
    prev = 1
    current = 1
    while count < n:
        current2 = prev + current
        prev = current 
        current = current2 
        count  += 1
    return current
```

```python
fib_loop(40)
```

```python
## count number of times, 5 shows up 

import random 

random_values = []
i = 0
while i < 100:
    random_values.append(random.randint(1, 20))
    i += 1

print(random_values)
```

```python
count_fives = 0

for val in random_values:
    if val == 5:
        count_fives += 1

print(count_fives)
```
